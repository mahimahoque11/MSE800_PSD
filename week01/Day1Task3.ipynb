```python
# input
num = int(input ("Please Enter A Number"))
p_input = int(input ("Please Enter The Power"))

result = 1
i = 1
pow_abs = abs(p_input)

if num == 0 and p_input < 0:
    print ("Zero cannot be a power of a negative number.")
else:
    while i <= pow_abs:
        result = result * num
        i += 1

if p_input < 0:
        result = 1 / result

print (num, "to the power of", p_input, "is", result)
```
